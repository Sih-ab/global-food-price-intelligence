```python
# Check dataset dimensions
print("Dataset shape:", df.shape)

# Check data types and non-null counts
df.info()

# Count missing values
print("\nColumns with missing values:")
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))
```

```python
missing_summary = df.isnull().sum()

missing_summary = missing_summary[missing_summary > 0].sort_values(
    ascending=False
)

print("Total columns with missing values:", len(missing_summary))
print("\nMissing values by column:")
print(missing_summary.to_string())

print("\nTotal missing cells:", df.isnull().sum().sum())
```

```python
# Show all rice-related columns
rice_columns = [col for col in df.columns if "rice" in col.lower()]

print("Rice-related columns:")
print(rice_columns)

# Check missing values in these columns
print("\nMissing values:")
print(df[rice_columns].isnull().sum().to_string())

# Preview rice-related data
print("\nSample data:")
print(df[rice_columns].head(10).to_string(index=False))
```

```python
# Compare rice-related variables
rice_price_columns = ["rice", "o_rice", "h_rice", "l_rice", "c_rice"]

print(df[rice_price_columns].describe().round(2))

print("\nNon-missing values:")
print(df[rice_price_columns].count())

print("\nFirst 10 observations:")
print(df[["price_date", "mkt_name"] + rice_price_columns].head(10).to_string(index=False))
```

```python
# Convert price_date to datetime
df["price_date"] = pd.to_datetime(df["price_date"])

# Check date range
print("Start date:", df["price_date"].min())
print("End date:", df["price_date"].max())

# Count unique markets
print("\nUnique markets:", df["mkt_name"].nunique())

# Check observations for each market
market_counts = df.groupby("mkt_name").size()

print("\nObservations per market:")
print(market_counts.describe().round(2))

# Check duplicate market-date combinations
duplicates = df.duplicated(
    subset=["mkt_name", "price_date"]
).sum()

print("\nDuplicate market-date rows:", duplicates)
```

```python
# Calculate monthly average rice price across all markets
monthly_rice = (
    df.groupby("price_date")["c_rice"]
    .mean()
    .reset_index()
)

print("Monthly rice price data:")
print(monthly_rice.head(10))

print("\nSummary statistics:")
print(monthly_rice["c_rice"].describe().round(2))

print("\nNumber of months:", len(monthly_rice))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_rice["price_date"],
    monthly_rice["c_rice"]
)

plt.title("Monthly Average Rice Price in Bangladesh (2007–2026)")
plt.xlabel("Date")
plt.ylabel("Average Rice Price")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Calculate yearly average rice price
yearly_rice = (
    df.groupby("year")["c_rice"]
    .mean()
    .reset_index()
)

print(yearly_rice.to_string(index=False))
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    yearly_rice["year"],
    yearly_rice["c_rice"],
    marker="o"
)

plt.title("Yearly Average Rice Price in Bangladesh")
plt.xlabel("Year")
plt.ylabel("Average Rice Price")
plt.xticks(yearly_rice["year"], rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Calculate month-to-month price change
monthly_rice = monthly_rice.sort_values("price_date").copy()

monthly_rice["price_change"] = monthly_rice["c_rice"].diff()

monthly_rice["price_change_pct"] = (
    monthly_rice["c_rice"].pct_change() * 100
)

print(
    monthly_rice[
        ["price_date", "c_rice", "price_change", "price_change_pct"]
    ].head(10).round(2).to_string(index=False)
)

print("\nLargest monthly increases:")
print(
    monthly_rice.nlargest(5, "price_change_pct")[
        ["price_date", "c_rice", "price_change_pct"]
    ].round(2).to_string(index=False)
)
```

```python
# Calculate IQR for monthly percentage changes
changes = monthly_rice["price_change_pct"].dropna()

Q1 = changes.quantile(0.25)
Q3 = changes.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential anomalies
anomalies = monthly_rice[
    (monthly_rice["price_change_pct"] < lower_bound) |
    (monthly_rice["price_change_pct"] > upper_bound)
].copy()

print("Lower bound:", round(lower_bound, 2), "%")
print("Upper bound:", round(upper_bound, 2), "%")
print("Potential anomalies:", len(anomalies))

print("\nDetected anomalies:")
print(
    anomalies[
        ["price_date", "c_rice", "price_change_pct"]
    ].round(2).to_string(index=False)
)
```

```python
plt.figure(figsize=(15, 6))

# Overall monthly rice price
plt.plot(
    monthly_rice["price_date"],
    monthly_rice["c_rice"],
    label="Average Rice Price"
)

# Potential anomalies
plt.scatter(
    anomalies["price_date"],
    anomalies["c_rice"],
    color="red",
    marker="o",
    s=60,
    label="Potential Anomaly"
)

plt.title("Rice Price Trend and Potential Anomalies in Bangladesh")
plt.xlabel("Date")
plt.ylabel("Average Rice Price")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Prepare monthly rice price data for forecasting

rice_ts = monthly_rice[
    ["price_date", "c_rice"]
].copy()

rice_ts = rice_ts.sort_values("price_date")
rice_ts = rice_ts.dropna(subset=["c_rice"])

rice_ts = rice_ts.set_index("price_date")

print("Dataset shape:", rice_ts.shape)

print("\nMissing values:")
print(rice_ts.isnull().sum())

print("\nFirst 5 rows:")
print(rice_ts.head())

print("\nLast 5 rows:")
print(rice_ts.tail())
```

```python
# Split time-series data chronologically

split_index = int(len(rice_ts) * 0.80)

train = rice_ts.iloc[:split_index]
test = rice_ts.iloc[split_index:]

print("Training observations:", len(train))
print("Testing observations:", len(test))

print("\nTraining period:")
print(train.index.min(), "to", train.index.max())

print("\nTesting period:")
print(test.index.min(), "to", test.index.max())
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Naive forecast: use the last training value for every test month
naive_forecast = np.repeat(train["c_rice"].iloc[-1], len(test))

# Evaluate the forecast
mae = mean_absolute_error(test["c_rice"], naive_forecast)
rmse = np.sqrt(mean_squared_error(test["c_rice"], naive_forecast))

print("Last training price:", round(train["c_rice"].iloc[-1], 2))
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    test.index,
    test["c_rice"],
    label="Actual Price"
)

plt.plot(
    test.index,
    naive_forecast,
    label="Naive Forecast",
    linestyle="--"
)

plt.title("Actual vs Naive Forecast — Rice Price")
plt.xlabel("Date")
plt.ylabel("Average Rice Price")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Train Holt's Exponential Smoothing model
model = ExponentialSmoothing(
    train["c_rice"],
    trend="add",
    seasonal=None,
    initialization_method="estimated"
)

fitted_model = model.fit()

# Forecast the test period
holt_forecast = fitted_model.forecast(len(test))

# Evaluate the forecast
mae_holt = mean_absolute_error(test["c_rice"], holt_forecast)
rmse_holt = np.sqrt(mean_squared_error(test["c_rice"], holt_forecast))

print("Holt Model Results")
print("MAE:", round(mae_holt, 2))
print("RMSE:", round(rmse_holt, 2))
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    test.index,
    test["c_rice"],
    label="Actual Price"
)

plt.plot(
    test.index,
    holt_forecast,
    linestyle="--",
    label="Holt Forecast"
)

plt.title("Actual vs Holt Forecast — Rice Price")
plt.xlabel("Date")
plt.ylabel("Average Rice Price")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Use the price from 12 months earlier
seasonal_forecast = rice_ts["c_rice"].shift(12).loc[test.index]

# Evaluate the forecast
mae_seasonal = mean_absolute_error(
    test["c_rice"],
    seasonal_forecast
)

rmse_seasonal = np.sqrt(
    mean_squared_error(test["c_rice"], seasonal_forecast)
)

print("Seasonal Naive Model Results")
print("MAE:", round(mae_seasonal, 2))
print("RMSE:", round(rmse_seasonal, 2))
```

```python
# Compare all forecasting models

results = pd.DataFrame({
    "Model": [
        "Naive Forecast",
        "Holt Exponential Smoothing",
        "Seasonal Naive"
    ],
    "MAE": [
        mae,
        mae_holt,
        mae_seasonal
    ],
    "RMSE": [
        rmse,
        rmse_holt,
        rmse_seasonal
    ]
})

results = results.sort_values("MAE")

print(results.round(3).to_string(index=False))
```

```python
# Forecast the next 12 months using the best baseline model

last_date = rice_ts.index.max()
last_price = rice_ts["c_rice"].iloc[-1]

future_dates = pd.date_range(
    start=last_date + pd.offsets.MonthBegin(1),
    periods=12,
    freq="MS"
)

future_forecast = pd.DataFrame({
    "price_date": future_dates,
    "forecast_c_rice": [last_price] * 12
})

print(future_forecast.to_string(index=False))
```

```python
plt.figure(figsize=(14, 6))

# Historical price
plt.plot(
    rice_ts.index,
    rice_ts["c_rice"],
    label="Historical Price"
)

# Future forecast
plt.plot(
    future_forecast["price_date"],
    future_forecast["forecast_c_rice"],
    linestyle="--",
    marker="o",
    label="12-Month Baseline Forecast"
)

plt.axvline(
    rice_ts.index.max(),
    linestyle=":",
    label="Forecast Start"
)

plt.title("Bangladesh Rice Price: Historical Data and Baseline Forecast")
plt.xlabel("Date")
plt.ylabel("Average Rice Price")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
